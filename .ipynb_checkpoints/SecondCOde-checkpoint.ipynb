```python
#recursion
def fact(n):
    if n==0 or n==1:
        return n
    else :
        return n*fact(n-1)
print(fact(2),fact(3),fact(4))
print("hi","hello","ha")
```

```python
def fib(n):
    if n==0 or n==1:
        return n
    else :
        return fib(n-1)+fib(n-2)
for i in range(9):
    print(fib(i),end=" ")
```

```python
def has_duplicates(L):
    ans=False
    for i in range(len(L)):
        if(L[i] in L[i+1:]):
            ans=True
            break
    return ans
has_duplicates([1,2,3,6,5,4,1])
```

```python
L=[11,23,22,42,25,7,62,99,87]
def mini(ind,L):
    if ind==0:
        return L[0]
    else:
        get=mini(ind-1,L)
        if(L[ind] < get ):
            return L[ind]
        else:
            return get
mini(len(L)-1,L)
```

```python
ui=["abcd","ab","abcde"]
ans=""

for r in range(min(len(x) for x in ui)):
    prif=ui[0][r]
    for i in range(1,len(ui)):
        if (ui[i][r]==prif):
            ans+=prif
print(ans)
```
