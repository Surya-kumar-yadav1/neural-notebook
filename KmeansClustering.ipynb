```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('Mall_Customers.csv')
```

```python
X=data.iloc[:,[3,4]].values
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1,11),wcss)
plt.xlabel("k")
plt.ylabel("wcss")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5,init="k-means++",random_state=42)
y=kmeans.fit_predict(X)
print(y)
```

```python
plt.scatter(X[y==0,0],X[y==0,1],s=100,c='pink',label='Cluster1')
plt.scatter(X[y==1,0],X[y==1,1],s=100,c='blue',label='Cluster2')
plt.scatter(X[y==2,0],X[y==2,1],s=100,c='cyan',label='Cluster3')
plt.scatter(X[y==3,0],X[y==3,1],s=100,c='orange',label='Cluster4')
plt.scatter(X[y==4,0],X[y==4,1],s=100,c='grey',label='Cluster5')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],s=200,c='red',label='Centroid')
plt.legend()
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```
