```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('Mall_Customers.csv')
```

```python
X=data.iloc[:,[3,4]].values 
```

```python
import scipy.cluster.hierarchy as sch
dendrogram = sch.dendrogram(sch.linkage(X,'ward'))
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
hc = AgglomerativeClustering(n_clusters=5,metric='euclidean',linkage='ward')
y=hc.fit_predict(X)
print(y)
```

```python
plt.scatter(X[y==0,0],X[y==0,1],s=100,c='pink',label='cluster 1')
plt.scatter(X[y==1,0],X[y==1,1],s=100,c='cyan',label='cluster 2')
plt.scatter(X[y==2,0],X[y==2,1],s=100,c='blue',label='cluster 3')
plt.scatter(X[y==3,0],X[y==3,1],s=100,c='green',label='cluster 4')
plt.scatter(X[y==4,0],X[y==4,1],s=100,c='yellow',label='cluster 5')
plt.legend()
plt.title('Agglomerative Clustering')
plt.ylabel('Spending Score')
plt.xlabel('Annual Income')
plt.show()
```
