```python
import matplotlib.pyplot as plt
hrsA=[1,2,3,4,5,6,7,8]
scoreA=[2,4,5,7,8,9,9,9]
hrsB=[0,0,0,9,10,16,7,18]
scoreB=[2,4,5,7,0,10,9,9]
plt.scatter(hrsA,scoreA,color='green',marker='*',label="Sec-A")
plt.scatter(hrsB,scoreB,color='blue',marker='o',label="Sec-B")
plt.title("Student's Data")
plt.ylabel("Score")
plt.xlabel("Hours")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

ax[0, 0].bar([1, 2, 3], [1, 4, 9],color='green')
ax[0, 0].set_title("Square")

ax[0, 1].hist([1, 2, 3], [1, 2, 3],color='orange')
ax[0, 1].set_title("Linear")

ax[1, 0].pie([1, 2, 3,5], labels=['A','B','C','D'],autopct="%1.1f%%")
ax[1, 0].set_title("Reverse")

ax[1, 1].scatter([1, 2, 3,7,8,9], [2, 4, 6,12,14,16],marker='*',color='red')
ax[1, 1].set_title("Double")

fig.suptitle("2×2 Subplots Example")
plt.tight_layout()
plt.savefig('graph.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(2, 2)

for i in range(2):
    for j in range(2):
        ax[i, j].plot([1*i, 2, 3+j], [1*i, 4, 9-j])
        ax[i, j].set_title("Very Long Title Example")
        ax[i, j].set_xlabel("X Axis Label")
        ax[i, j].set_ylabel("Y Axis Label")

plt.tight_layout()
plt.show()
```
